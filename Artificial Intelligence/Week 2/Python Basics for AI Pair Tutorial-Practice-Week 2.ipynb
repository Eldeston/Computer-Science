# Station 3
Conditionals let code make decisions. An AI model gives a number; an if statement turns that number into an action. Python checks the conditions from top to bottom and runs only the first block that is True. Note the colon : at the end of each line and the 4 space indent below it.

| Operator     | Meaning                     | Example                                 |
|--------------|-----------------------------|-----------------------------------------|
| `==`         | Equal to                    | `label == "cat"`                        |
| `!=`         | Not equal to                | `label != "cat"`                        |
| `> < >= <=`  | Comparisons                 | `score >= 0.5`                          |
| `and`        | Both must be `True`         | `score > 0.5 and label == "cat"`        |
| `or`         | At least one `True`         | `genre == "jazz" or genre == "blues"`   |
| `not`        | Reverses `True` and `False` | `not is_trained`                        |

```python
confidence = 0.82

if confidence >= 0.9:
    print("Very sure: label the image automatically")
elif confidence >= 0.6:
    print("Fairly sure: ask a human to check")
else:
    print("Not sure: reject the prediction")
```

# Station 4
A for loop repeats code once for every item. Training a model is a loop: for every epoch, for every batch, learn a little.

| You write          | Python gives     |
|--------------------|------------------|
| `range(5)`         | 0, 1, 2, 3, 4    |
| `range(2, 6)`      | 2, 3, 4, 5       |
| `range(0, 10, 2)`  | 0, 2, 4, 6, 8    |

```python
colours = ["coral", "teal", "gold"]

for colour in colours:                 # loop over items
    print("Painting with", colour)

for epoch in range(1, 4):              # range(1, 4) gives 1, 2, 3
    print(f"Epoch {epoch} of 3")

for i, colour in enumerate(colours):   # index and item together
    print(i, colour)

squares = [n * n for n in range(5)]    # list comprehension: a loop in one line
print(squares)                         # [0, 1, 4, 9, 16]
```

## Exercise 4.1 (10 Points)
Simulate a model learning. The loss falls a little every epoch.

```python
loss = 2.0
history = []

for epoch in range(1, 6):                       # epochs 1 to 5
    loss = loss * 0.7                          # the model improves by 30%
    history.append(round(loss, 3))               # save the loss
    print(f"Epoch {epoch}: loss = {loss:.3f}")

print("Loss history:", history)

# Count how many epochs had loss below 1.0
good_epochs = 0
for value in history:
    if value < 1.0:
        good_epochs = good_epochs + 1
print("Epochs with loss below 1.0:", good_epochs)

# The same count with a list comprehension
print(len([v for v in history if v < 1.0]))
```

# Station 5
A dictionary stores pairs of key: value, inside curly brackets. Use it when you want to look something up by name instead of by position. Model settings, a row of data and a JSON reply from an AI API are all dictionaries.

| Operation         | Code                       |
|-------------------|----------------------------|
| All keys          | `artwork.keys()`           |
| All values        | `artwork.values()`         |
| Is a key there?   | `"title" in artwork`       |
| Remove a pair     | `del artwork["price_aed"]` |

```python
artwork = {
    "title": "Desert Dawn",
    "artist": "Mariam",
    "year": 2026,
    "styles": ["digital", "calligraphy"],     # a value can be a list
}

print(artwork["title"])                 # look up by key
artwork["price_aed"] = 1500              # add a new pair
artwork["year"] = 2025                   # change a value
print(artwork.get("medium", "unknown"))  # safe lookup with a default

for key, value in artwork.items():       # loop over every pair
    print(key, "->", value)
```

## Exercise 6.1 (10 Points)
Write three small functions that AI projects really use.

```python
def accuracy(correct, total):
    """Return accuracy as a percentage."""
    return correct / total * 100

def normalise(values):
    """Scale numbers to the range 0 to 1."""
    low, high = min(values), max(values)
    return [(v - low) / (high - low) for v in values]

def label_from_score(score, threshold=0.5):
    """Turn a model's score into a label."""
    if score >= threshold:
        return "positive"
    return "negative"

print(accuracy(45, 50))                        # 90.0
print(normalise([10, 20, 30, 50]))             # [0.0, 0.25, 0.5, 1.0]
print(label_from_score(0.73))                  # positive
print(label_from_score(0.73, threshold=0.75))  # make this print negative
```

# Station 7
NumPy gives Python fast arrays: grids of numbers you can calculate with in one line, without a loop. Images, sound and every neural network weight are NumPy style arrays.

| Code | What it does |
|------|--------------|
| `np.array(list)` | Makes an array from a list |
| `a.shape` | Size of each dimension, e.g. `(28, 28)` |
| `np.zeros((r, c))`, `np.ones((r, c))` | Arrays full of `0` or `1` |
| `np.random.rand(r, c)` | Random numbers between `0` and `1` |
| `a.mean()`, `a.sum()`, `a.max()` | Summary statistics |
| `a[a > 5]` | Filter with a condition |
| `a.reshape(r, c)` | Change the shape, same numbers |

```python
import numpy as np

prices = [10, 20, 30]
# prices * 2 on a list repeats it: [10, 20, 30, 10, 20, 30]

arr = np.array([10, 20, 30])
print(arr * 2)                 # [20 40 60]  maths on every item
print(arr.mean(), arr.max())   # 20.0 30

grid = np.zeros((3, 4))        # 3 rows, 4 columns of zeros
print(grid.shape)              # (3, 4)
print(np.arange(0, 10, 2))     # [0 2 4 6 8]
print(arr[arr > 15])           # [20 30]  keep only items above 15
```

## Exercise 8.1 (12 Points)
Explore, clean and question the song data.

```python
import pandas as pd

songs = pd.DataFrame({
    "title":  ["Oud Nights", "Desert Wind", "Neon Souq", "Blue Dhow", "Pearl", "Sandstorm"],
    "genre":  ["khaleeji", "ambient", "pop", "jazz", "khaleeji", "pop"],
    "tempo":  [96, 70, 128, 110, None, 140],
    "streams_k": [540, 120, 980, 75, 610, 870],
})

songs.head()   

print("Rows and columns:", songs.shape)
print(songs["streams_k"].mean())                    # average streams

print("Missing values:")
print(songs.isna().sum())
songs["tempo"] = songs["tempo"].fillna(songs["tempo"].median())   # fill with the median

fast = songs[songs["tempo"] > 120]                # songs faster than 120 bpm
print(fast[["title", "tempo"]])

top = songs.sort_values("streams_k", ascending=False)  # most streamed first
print("Top song:", top.iloc[0]["title"])

songs["popular"] = songs["streams_k"] > 500           # a new column: a label for AI!
print(songs.groupby("genre")["streams_k"].mean())

songs["streams_k"].plot(kind="bar", title="Streams (thousands)")
```
