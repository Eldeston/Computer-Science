## 🟢 Core · Step 4 — A palette that matches your theme
Set the **minimum and maximum** for red, green and blue (0–255). Cool/ocean → high blue. Earthy → medium red and green, low blue. Neon → push one channel to 255.

```python
import random
import matplotlib.pyplot as plt

def theme_colour():
    r = random.randint(0, 255)
    g = random.randint(0, 255)
    b = random.randint(0, 255)
    return f"#{r:02X}{g:02X}{b:02X}"

palette = [theme_colour() for _ in range(5)]

fig, ax = plt.subplots(figsize=(8, 2))

for i, c in enumerate(palette):
    ax.add_patch(plt.Rectangle((i, 0), 1, 1, color=c))
    ax.text(i + 0.5, -0.15, c, ha="center", va="top")

ax.set_xlim(0, 5); ax.set_ylim(-0.4, 1); ax.axis("off")
plt.show()
```

## 🟢 Core · Step 5 — Make a poster for the gallery walk

```python
import textwrap

ideas = [f"{random.choice(part_a)} {random.choice(part_b)} {random.choice(part_c)}" for _ in range(4)]

fig, ax = plt.subplots(figsize=(6, 8), facecolor=palette[0])
ax.set_facecolor(palette[0])

for i, c in enumerate(palette[1:]):
    ax.add_patch(plt.Rectangle((i * 0.25, 0), 0.25, 0.18, color=c))

ax.text(0.08, 0.92, theme.upper(), fontsize=11, color="white")
y = 0.82

for line in ideas:
    for part in textwrap.wrap(line, 24):          # wrap long lines so they fit
        ax.text(0.08, y, part, fontsize=15, color="white", fontweight="bold", family="serif")
        y -= 0.06
    y -= 0.03

ax.set_xlim(0, 1); ax.set_ylim(0, 1); ax.set_xticks([]); ax.set_yticks([])
fig.savefig("week1_poster.png", dpi=150, bbox_inches="tight")
plt.show()
```

## 🔵 Stretch · A mood-driven generator
Goal: the user types a mood (`calm`, `angry`, `joyful`…) and gets words **and** colours that match.

**Hints**
1. Make a **dictionary** that maps each mood to its own word list and colour ranges.
2. Write a function `generate(mood)` that returns a poem and a palette.
3. Use a `while` loop with `input()` so the user can keep asking until they type `quit`.

```python
import random

moods = {
    "calm": {
        "words": ["still water", "soft linen", "a slow tide", "morning fog", "a quiet room"],
        "rgb":   [(80, 160), (140, 210), (180, 255)]      # (min, max) for R, G, B
    },
    "angry": {
        "words": ["violent storm", "crimson flames", "exploding lava", "breaking glass", "roaring lion"],
        "rgb":   [(180, 255), (140, 210), (80, 160)]      # (min, max) for R, G, B
    },
    "joyful": {
        "words": ["golden sunlight", "laughter", "spring flowers", "bright skies", "celebration"],
        "rgb": [(180, 255), (180, 255), (60, 180)]
    },

    "sad": {
        "words": ["falling rain", "empty streets", "grey clouds", "lonely echoes", "winter dusk"],
        "rgb": [(50, 120), (50, 120), (100, 180)]
    },
    "fearful": {
        "words": ["creaking floorboards", "unseen eyes", "dark forests", "distant footsteps", "cold silence"],
        "rgb": [(20, 80), (20, 80), (20, 120)]
    }
    # add at least two more moods here: "angry", "joyful", ...
}

def generateMood(mood):
    baseMood = moods[mood]

    rgbList = baseMood["rgb"]
    wordsList = baseMood["words"]
    
    wordsCount = len(wordsList)

    wordsRand = random.randint(0, max(wordsCount - 1, 0))
    redRand = random.randint(rgbList[0][0], rgbList[0][1])
    greenRand = random.randint(rgbList[1][0], rgbList[1][1])
    blueRand = random.randint(rgbList[2][0], rgbList[2][1])

    return (wordsList[wordsRand], f"#{redRand:02X}{greenRand:02X}{blueRand:02X}")

    # return  # build and return (poem_lines, palette)

# palette = [generateMood("angry") for _ in range(5)]
palette = [generateMood(mood) for mood in moods]

fig, ax = plt.subplots(figsize=(8, 2))

for i, (word, color) in enumerate(palette):
    ax.add_patch(plt.Rectangle((i, 0), 1, 1, color=color))
    ax.text(i + 0.5, -0.15, word, ha="center", va="top")
    ax.text(i + 0.5, -0.30, color, ha="center", va="top")

ax.set_xlim(0, 5); ax.set_ylim(-0.4, 1); ax.axis("off")
plt.show()

# while True:
#     mood = input("Mood (or 'quit'): ").strip().lower()
#     ...
```
